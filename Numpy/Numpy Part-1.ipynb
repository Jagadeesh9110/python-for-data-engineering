## Chapter 1: Why NumPy?

### 🔹 The Problem: Python Lists

Python lists are **general-purpose containers** — designed to store *any* type of Python object.

```python
data = [10, "Jagadeesh", True, 98.5, [1, 2, 3], {"age": 22}]
```

Since a list can hold mixed types, it doesn't store the actual values directly.
Instead, it stores **references (memory addresses)** pointing to separate Python objects.

**Internally, it looks like this:**

```
Python List
    │
    ├──► Integer Object (10)
    ├──► String Object ("Hello")
    ├──► Float Object (98.5)
    └──► Boolean Object (True)
```

> 💡 **Key idea:** A Python list stores *references* to objects, not the objects themselves.

---

### 🔹 Why Does Python Do This?

Because lists are built to hold **heterogeneous (mixed-type) data**:

```python
employee = [
    101,
    "Rahul",
    True,
    75000.50,
    {"department": "IT"}
]
```

Different data types coexisting in one list — that flexibility is *why* references are needed.

---

### 🔹 The Drawback

For large-scale numerical work (millions of values):

- Every element is a separate Python object.
- The list only stores references to them.
- Python processes elements one at a time.
- Result → high memory usage + high computational overhead.

---

### 🔹 The Solution: NumPy

NumPy was built for **large-scale numerical computing**.

Instead of scattering references to separate objects, NumPy stores elements of the **same data type** in one **contiguous block of memory**:

```
50000 | 60000 | 70000 | 80000 | 90000
```

This layout is far more memory-efficient and enables much faster computation.

> ### 🧠 First Principle of NumPy
> **A NumPy array stores elements of the same data type in contiguous memory.**
>
> This single idea is the foundation for everything else in NumPy.

---

### 🔹 Why Is NumPy Faster?

| Reason | Explanation |
|---|---|
| Homogeneous data | All elements share one data type |
| Contiguous memory | Data sits together in memory, not scattered |
| Less object overhead | No per-element Python object wrapping |
| C-level execution | Operations run via optimized, compiled C code |
| No Python-level loops | Loops written by the user aren't needed |

> ⚠️ **Important clarification:** NumPy does **NOT** eliminate looping altogether.
> It eliminates **Python-level** loops. Internally, NumPy still iterates over data — just using fast, compiled C code instead of slow Python bytecode.

---

### 🔹 When Should You Use NumPy?

Use NumPy when:
- ✅ Data is numerical
- ✅ All elements share the same data type
- ✅ Dataset is large
- ✅ High-performance math/computation is needed

**Real-world examples:**
- Employee salaries
- Attendance percentages
- Sales data
- Sensor readings
- Financial datasets
- Machine Learning datasets

---

### 🔑 Key Takeaways — Chapter 1

- Python lists store **references** to Python objects.
- Python lists support **heterogeneous** data.
- NumPy arrays store **homogeneous** data.
- NumPy stores data in **contiguous memory**.
- NumPy is optimized for **numerical computation**.
- NumPy removes **Python-level iteration overhead** via compiled C code (not looping itself).

# NumPy - Chapter 2: ndarray (N-Dimensional Array)

## What is ndarray?

`ndarray` stands for **N-Dimensional Array**.

The **N** represents **any number of dimensions**.

NumPy uses the same data structure to represent:

- 0D (Scalar)
- 1D (Vector / List)
- 2D (Matrix / Table)
- 3D (Collection of Tables)
- ND (Higher-dimensional arrays)

---

## Dimension Hierarchy

### 0D

A single value.

Example:

```python
50000
```

---

### 1D

A sequence of values.

Example:

```python
[50000, 60000, 70000]
```

---

### 2D

Rows and columns.

Example:

```python
[
    [101, 50000],
    [102, 60000],
    [103, 70000]
]
```

---

### 3D

A collection of multiple 2D arrays.

Example:

Monthly employee salary tables.

---

## Why One Class?

NumPy uses a single class:

```python
numpy.ndarray
```

instead of creating separate classes like:

- Array1D
- Array2D
- Array3D

This provides a consistent API regardless of the number of dimensions.

---

## Data Engineering Examples

- Employee IDs → 1D
- Salary Table → 2D
- Monthly Salary Data → 3D
- Multi-year Department Data → ND

---

## Key Takeaways

- `ndarray` = N-Dimensional Array.

```

**▶️ Code Cell — Boolean Array**

bool
```

**▶️ Code Cell — String Array**

---
 
### 🔹 Mixed Data & Type Promotion
 
**📝 Concept**
When an array is created from mixed types, NumPy must pick **one** common dtype for the whole array — since it can only store one type at a time.
 
**▶️ Code Cell — Int + Float**
```python
arr = np.array([10, 20.5, 30])
 
print(arr)
print(arr.dtype)
```
 
**✅ Output** *(verified)*
```
[10.  20.5 30. ]
float64
```
 
💡 **Why?** Since one value is a float, every integer is **promoted to float** — NumPy can't mix types, so it upgrades everything to the "widest" common type.
 
---
 
**▶️ Code Cell — Int + Boolean**
```python
arr = np.array([10, True, 30])
 
print(arr)
print(arr.dtype)
```
 
**✅ Output** *(verified)*
```
[10  1 30]
int64
```
 
💡 **Why?** Internally, `True = 1` and `False = 0`. NumPy converts the boolean into an integer, so everything becomes `int64`.
 
---
 
**▶️ Code Cell — Int + String**
```python
arr = np.array([10, "20", 30])
 
print(arr)
print(arr.dtype)
```
 
**✅ Output** *(verified)*
```
['10' '20' '30']
<U21
```
 
💡 **Why?** A string can represent anything (numbers, text, etc.), but a number **cannot** represent arbitrary text. So NumPy converts *everything* to strings — the safest common type.
 
---
 
### 🔑 Rule — Type Promotion
 
> **NumPy always tries to find one common data type that can represent every element in the array.**
> This process is called **Type Promotion**.
 
**General hierarchy (weakest → strongest):** `bool → int → float → string`
(Whichever type appears, everything gets promoted *up* to that type — never down.)
 
---
 
### Key Takeaways — Chapter 3
 
- `.dtype` tells you the single fixed data type an array is using internally.
- Numeric arrays default to `int64` or `float64` on most 64-bit systems.
- String arrays show up as `<U#` (Unicode string of max length `#`).
- Mixing types triggers **Type Promotion** — NumPy converts everything to one common type:
  - `int + float → float`
  - `int + bool → int` (since `True/False` = `1/0`)
  - `int + string → string` (strings win over everything, since they're the most "general" type)

| dtype         | Meaning        |
| ------------- | -------------- |
| int8          | 8-bit Integer  |
| int16         | 16-bit Integer |
| int32         | 32-bit Integer |
| int64         | 64-bit Integer |
| float32       | 32-bit Float   |
| float64       | 64-bit Float   |
| bool          | Boolean        |
| str / Unicode | String         |

# Precision:

Precision refers to the level of detail and accuracy with which numerical values are represented. NumPy data types have different levels of precision, which determine the range of values they can represent and the amount of memory they occupy.

````python
 # Create an array with different data types
arr_int32 = np.array([1234567890, 1234567890], dtype=np.int32)
arr_int64 = np.array([1234567890, 1234567890], dtype=np.int64)

print("Data type of arr_int32:", arr_int32.dtype)   # Output: int32
print("Data type of arr_int64:", arr_int64.dtype)   # Output: int64

print(arr_int32)

# In this example, the `int32` data type has limited precision compared to `int64`, which can represent larger integers without loss of precision.
````

## **Impact of Precision on Memory Usage:**

````python
# Create arrays with different data types
arr_float32 = np.array([1.1, 2.2, 3.3], dtype=np.float32)
arr_float64 = np.array([1.1, 2.2, 3.3], dtype=np.float64)

print("Memory usage of arr_float32:", arr_float32.itemsize * arr_float32.size, "bytes")  # Output: 12 bytes (3 elements * 4 bytes/element)
print("Memory usage of arr_float64:", arr_float64.itemsize * arr_float64.size, "bytes")  # Output: 24 bytes (3 elements * 8 bytes/element)
# In this 32 bit is less precise and 64 bit  is more precise
````

````python
# Create a large array with float32 and float64 data types
large_array_float32 = np.arange(1000000, dtype=np.float32)
large_array_float64 = np.arange(1000000, dtype=np.float64)

# Calculate the sum of elements
sum_float32 = np.sum(large_array_float32)
sum_float64 = np.sum(large_array_float64)

print("Sum using float32:", sum_float32)
print("Sum using float64:", sum_float64) # this will is correct because higher precise will have the correct value it will have more space
````

### 3. `np.empty()`
 
This function is commonly misunderstood. It does **not** create an array filled with zeros — it allocates memory without initializing the values.
 
```python
arr = np.empty((3, 3))
 
print(arr)
```
 
Output (verified on this system):
```
[[3.62734033e-315 0.00000000e+000 6.94012680e-310]
 [5.83038173e-317 6.94012680e-310 6.94012680e-310]
 [5.83661486e-317 3.62550786e-315 3.95252517e-322]]
```
 
The exact values will differ across machines and runs, since they are leftover data from whatever was previously stored in that memory location.
 
**Why use it?**
 
Initializing millions of values to zero takes time. If every value in the array is going to be overwritten immediately after creation, `np.empty()` skips that unnecessary initialization step and is faster.
 
---

Suppose every employee receives a fixed joining bonus of 5000. Rather than writing this out manually:

---
 
### Summary — Chapter 5
 
| Function | Fills array with | Default dtype |
|---|---|---|
| `np.zeros(shape)` | `0` | `float64` |

| `np.full(shape, value)` | Given `value` | Matches type of `value` |
 
`np.empty()` trades predictability for speed — use it only when every element will be overwritten before being read.

### 2. `np.linspace()`
 
One of the more misunderstood functions. Unlike `arange()`, where you specify the step size, `linspace()` requires you to specify how many values you want in total.
 
**Syntax:**
```python
np.linspace(start, stop, number_of_values)
```
 
**Example:**
```python
arr = np.linspace(0, 10, 5)
print(arr)
```
 
Output (verified):
```
[ 0.   2.5  5.   7.5 10. ]
```
 
Exactly 5 values are returned, evenly spaced between 0 and 10 inclusive.
 
**Another example:**
```python
arr = np.linspace(100, 200, 11)
print(arr)
```
 
Output (verified):
```
[100. 110. 120. 130. 140. 150. 160. 170. 180. 190. 200.]
```
 
**Key difference:**
 
- `np.arange(0, 10, 2)` — gives values spaced 2 units apart.
- `np.linspace(0, 10, 6)` — divides the interval `[0, 10]` into 6 equally spaced values.
**When to use it:** most common in Machine Learning, simulations, graph plotting, and scientific computing. Less commonly used in ETL pipelines, where `arange()` is more typical.

### 6. `np.asarray()`
 
Commonly used internally by other libraries.

**Key difference:**

## Understanding NumPy 2D Array Slicing

When you have a 2D NumPy array (rows and columns), you can access specific rows, columns, or sub-blocks using this general syntax:

    array[row_selector, column_selector]

- The part **before the comma** controls which **rows** you select.
- The part **after the comma** controls which **columns** you select.
- A colon `:` alone means "give me everything" (all rows, or all columns).

### Common patterns

| Syntax | Meaning |
|---|---|
| `array[:, 0]` | All rows, only column 0 |
| `array[:, 1]` | All rows, only column 1 |
| `array[:2, :]` | Rows 0 and 1 (index 2 excluded), all columns |
| `array[:2, 0]` | Rows 0 and 1, only column 0 |
| `array[-3:, :]` | Last 3 rows, all columns |
| `array[::2]` | Every 2nd row (step = 2), all columns |
| `array[::-1]` | All rows, reversed order |
| `array[1:4, :]` | Rows 1, 2, 3 (index 4 excluded), all columns |
| `array[0:2, 0:2]` | Rows 0-1 AND columns 0-1 → a sub-block |
| `array[1:, 1:]` | All rows from index 1 onward AND all columns from index 1 onward |

### Key rules to remember

1. **Slicing is `start:stop:step`** — `stop` is always **excluded**. So `[1:4]` gives indices 1, 2, 3 — not 4.
2. **Negative indices count from the end.** `-1` is the last row, `-3` is the third-from-last row.
3. **Leaving a number blank** means "go to the start" or "go to the end" — e.g. `[2:]` means "from index 2 to the end", `[:3]` means "from the start up to (not including) index 3".
4. **A single column index (like `[:, 0]`)** returns a flattened 1D array, while a **range (like `[:, 0:1]`)** keeps it as a 2D column — this is an important distinction to notice.
5. Combining a row slice and a column slice at the same time (`array[row_slice, col_slice]`) lets you pull out a **rectangular sub-matrix** in one step.

Once you're comfortable with this pattern, all the slicing examples below will make a lot more sense.

# NumPy - Boolean Indexing: Empty Arrays

## Does Boolean Indexing Raise an Error if No Elements Match?

**No.**

Boolean indexing **does not raise an error** if no elements satisfy the condition.

Instead, NumPy returns an **empty ndarray**.

---

## Example

```python
import numpy as np

attendance = np.array([95, 88, 97, 91, 84, 99])

invalid = attendance[attendance < 80]

print(invalid)
```

### Output

```python
[]
```

---

## Why?

First, NumPy evaluates the condition.

```python
attendance < 80
```

Result:

```python
[False False False False False False]
```

Since every value is `False`, there are **no matching elements**.

NumPy simply returns an empty array.

```python
[]
```

No exception is raised.

---

## Type of the Result

```python
print(type(invalid))
```

Output

```python
<class 'numpy.ndarray'>
```

Even though it is empty, it is still a valid NumPy array.

---

## Size of the Result

```python
print(invalid.size)
```

Output

```python
0
```

or

```python
print(len(invalid))
```

Output

```python
0
```

---

## Why Doesn't NumPy Raise an Error?

Boolean indexing is a **filtering operation**, similar to SQL.

SQL Example:

```sql
SELECT *
FROM employees
WHERE salary > 1000000;
```

If no employee satisfies the condition, SQL returns:

```
0 Rows
```

It does **not** raise an error.

NumPy follows the same behavior.

---

## When Does an Error Occur?

An error occurs **only when you try to access an element that doesn't exist.**

Example:

```python
invalid = attendance[attendance < 80]

print(invalid)
```

Output

```python
[]
```

No error.

But,

```python
print(invalid[0])
```

Output

```text
IndexError: index 0 is out of bounds for axis 0 with size 0
```

Reason:

The array contains **zero elements**, so index `0` does not exist.

---

## Safe Way

Always check whether the array contains any elements before accessing them.

```python
if invalid.size > 0:
    print(invalid[0])
else:
    print("No invalid attendance records found.")
```

Output

```text
No invalid attendance records found.
```

---

## Real Data Engineering Example

```python
invalid_salary = salary[salary < 0]
```

If there are no negative salaries,

```python
print(invalid_salary)
```

Output

```python
[]
```

This simply means the dataset contains **no invalid salary records**, and the ETL pipeline can continue safely.

---

## Key Takeaways

- Boolean indexing returns a **NumPy array**.
- If no elements match the condition, NumPy returns an **empty ndarray (`[]`)**.
- No exception is raised during filtering.
- An `IndexError` occurs **only when accessing an element** from an empty array.
- Always check `array.size > 0` before accessing elements if the result may be empty.
- This behavior makes Boolean indexing ideal for filtering operations in ETL and data engineering pipelines.

### Advanced Fancy Indexing

Suppose you want

    employees[0,1]
    employees[2,1]
    employees[5,1]

Instead of writing three separate statements,
use Fancy Indexing.

````python
rows = [0,2,5]

cols = [1,1,1]

print(employees[rows, cols])
````

Output

    [50000 71000 90000]

Explanation
NumPy pairs the indices element by element.

    (0,1)
    (2,1)
    (5,1)
