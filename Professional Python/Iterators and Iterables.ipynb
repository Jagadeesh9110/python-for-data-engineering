# Iterators and Iterables

Almost everything that can be used in a `for` loop in Python is an **iterable**.

Examples:

- list
- tuple
- string
- dictionary
- set
- range
- file objects

Python processes all of them using a common mechanism called the **Iterator Protocol**.

Understanding iterators helps explain how Python loops actually work internally.

## Why Learn This?

Consider this basic list iteration code:

```python
numbers = [10, 20, 30]

for num in numbers:
    print(num)
```

**Output**:
```text
10
20
30
```

This syntax is incredibly simple on the surface. But have you ever wondered: **how does the `for` loop actually know which element comes next?**

The truth is, Python does not directly access the list items behind the scenes. Instead, it creates an intermediate object called an **iterator** to systematically retrieve the data.


| Term     | Meaning                                 |
| -------- | --------------------------------------- |
| Iterable | An object you can iterate over          |
| Iterator | An object that gives one item at a time |


## What is an Iterable?

An **iterable** is any object that can produce its elements one by one. It acts as a data container that Python can loop over. 

When you pass an iterable to the built-in `iter()` function, Python creates an underlying **iterator** engine for it.


### 1. List Iterable
Lists are the most common sequential iterables used in Python data structures.


````python
numbers = [10, 20, 30]

# Verify it is an iterable by creating an iterator object from it
iterator_obj = iter(numbers)
print("Type confirmation:", type(iterator_obj))

for num in numbers:
    print("Value:", num)
````

### 2. String Iterable
Strings are sequences of individual character elements, making them natively iterable.


````python
text = "Python"

# Loop over the character items
for character in text:
    print("Character:", character)
````

### 3. Dictionary Iterable
Dictionaries iterate through their **keys** by default, but you can also traverse their values or key-value pairs.


````python
student = {
    "name": "Alice",
    "age": 22
}

print("--- Iterating Default Keys ---")
for key in student:
    print("Key:", key)

print("\n--- Iterating Key-Value Pairs ---")
for key, value in student.items():
    print(f"Key: {key:4} | Value: {value}")
````

### 4. Set Iterable
Sets are iterables containing distinct, unordered items.


````python
colors = {"red", "blue", "green"}

for color in colors:
    print("Color item:", color)
````

### 5. Range Iterable
The `range()` object is a memory-efficient iterable that computes integer sequences lazily on demand.


````python
# Iterating over a lazy range generation stream
for index in range(5):
    print("Sequence Index:", index)
````

## What is an Iterator?

An **iterator** is an agent object that tracks traversal state behind the scenes. It explicitly remembers:
* **The current position** within the underlying dataset container.
* **The next element** scheduled to be returned to the execution context.

An iterator acts like a temporary **bookmark** placed inside a massive book. It marks your exact page location without altering the text of the book itself.


## Creating an Iterator

To generate an iterator manually from any standard data container, pass the collection container straight to Python's built-in **`iter()`** engine function.


````python
# Instantiate a standard list iterable
numbers = [10, 20, 30]

# Generate a standalone iterator object instance
it = iter(numbers)

print("--- Inspecting the Iterator ---")
print("Raw Object Pointer:", it)
print("Is the iterator identical to the list?", it is numbers)
````

## Getting Elements Natively

To retrieve items from an iterator instance one by one, pass the tracking agent to Python's built-in **`next()`** invocation function.

### Sequential Forward Traversal
```python
numbers = [10, 20, 30]
it = iter(numbers)

print(next(it))  # Output: 10
print(next(it))  # Output: 20
print(next(it))  # Output: 30
```

Every time you execute `next()`, the bookmark moves forward by exactly one element position inside the dataset container.


## The `StopIteration` Boundary Exception

If you request an additional element *after* reaching the end of the collection, Python raises a **`StopIteration`** exception:
```python
print(next(it))  # Raises: StopIteration
```

### Internal Traversal View
```text
Initial State:          [10]   20    30  (Iterator points to index 0)
After next(it) -> 10:    10   [20]   30
After next(it) -> 20:    10    20   [30]
After next(it) -> 30:    10    20    30  ──► [End Boundary Reached]

Subsequent next(it):    Raises StopIteration immediately
```


## How Does a `for` Loop Work Behind the Scenes?

When Python encounters a standard loop syntax, it transforms the code under the hood into an explicit `while` loop that relies on the structural iterator protocol:

```python
# What you write
for num in numbers:
    print(num)

# What Python executes internally
it = iter(numbers)
while True:
    try:
        num = next(it)
        print(num)
    except StopIteration:
        break  # Safely exit the infinite loop frame when data runs out
```
This is the foundational logic driving every single `for` loop iteration inside the Python runtime environment.


````python
# Verifying the internal while-loop iteration pattern
numbers_list = [10, 20, 30]
iterator_agent = iter(numbers_list)

print("--- Simulating a for loop using next() and try-except ---")
while True:
    try:
        element = next(iterator_agent)
        print("Processed Item:", element)
    except StopIteration:
        print("Boundary exception intercepted cleanly. Halting loop execution.")
        break
````

## Iteration Profiles Across Data Containers

### 1. String Iteration Mechanics
Strings process characters sequentially until hitting the string boundary:
```python
text = "CAT"
it = iter(text)
print(next(it))  # 'C'
print(next(it))  # 'A'
print(next(it))  # 'T'
print(next(it))  # Raises StopIteration
```

### 2. Dictionary Iteration Mechanics
By default, invoking an iterator across a dictionary container fetches its explicit **keys**:
```python
student = {"name": "Alice", "age": 22}
it = iter(student)
print(next(it))  # Output: "name"
print(next(it))  # Output: "age"
```


## Memory-Efficient Stream Processing

Files are natively integrated into the iterator protocol. When processing textual data logs from storage disks, the open file pointer serves data line by line on-demand:

```python
with open("employees.txt") as file:
    for line in file:
        print(line)
```

### The 50 GB Scale Problem
Imagine managing an analytics pipeline tasked with processing a **50 GB data log file**. Attempting to load the entire document into system memory at once would instantly cause an out-of-memory crash. 

Iterators bypass this storage limit by maintaining an on-demand lifecycle stream:

```text
[ Read Line 1 ] ──► Process Logic ──► Discard Memory Block
[ Read Line 2 ] ──► Process Logic ──► Discard Memory Block
```

Only one single line string object is retained in your system memory cache at any given fraction of a second.


## Core Concept Distinctions

| Iterable | Iterator |
| :--- | :--- |
| Primarily **stores** the raw data items. | Actively **traverses** and indexes target elements. |
| Responsibly creates a fresh iterator object. | Explicitly **remembers** its unique index location. |
| Interacts directly with the `iter()` function. | Interacts directly with the `next()` function. |
| **Example**: `list`, `str`, `dict`, `set` | **Example**: `list_iterator`, `str_iterator` |


````python
# Programmatically checking container iteration outputs
mock_dict = {"id": "EMP101", "role": "Data Engineer"}
dict_iterator = iter(mock_dict)

print("--- Testing Dictionary Key Streaming ---")
print("First token extracted: ", next(dict_iterator))
print("Second token extracted:", next(dict_iterator))

# Setup temporary file simulation for streaming check
with open("stream_test.txt", "w") as f:
    f.write("Line Item Alpha\nLine Item Beta\n")

print("\n--- Testing Streaming File Line Reading ---")
with open("stream_test.txt", "r") as file_stream:
    # Fetching lines one at a time using next() directly on the file object
    print("Chunk A:", repr(next(file_stream)))
    print("Chunk B:", repr(next(file_stream)))

# Housekeeping: Cleanup file system references
import os
if os.path.exists("stream_test.txt"):
    os.remove("stream_test.txt")
````
