# `timedelta`

A `timedelta` object represents the difference (duration) between two dates or times.

Examples:

- 5 days
- 10 hours
- 30 minutes
- 2 weeks

Unlike `date` or `datetime`, a `timedelta` does **not** represent a point in time.

Instead, it represents an amount of time.

## Creating a timedelta

````python
from datetime import timedelta

delta = timedelta(days=5)

print(delta)
````

## You can specify
```python
timedelta(
    days=5,
    hours=3,
    minutes=20,
    seconds=15
)
```

````python
from datetime import timedelta

delta = timedelta(
    days=5,
    hours=3,
    minutes=20,
    seconds=15
)

print(delta)
````

````python
delta = timedelta(
    days=7,
    weeks=2,
    hours=5,
    minutes=30,
    seconds=20,
    milliseconds=100,
    microseconds=500
)
print(delta)
````

````python
last_week = today - timedelta(days=7)

print(last_week)
````

## Comparing Dates

- Dates support all comparison operators.

````python
date1 = date(2026,7,26)

date2 = date(2026,8,1)

print(date1 < date2)

print(date1 > date2)

print(date1 == date2)
````

### The Evaluation Rules

* **Chronological Sorting**: Python evaluates date objects chronologically. An earlier calendar date is considered less than (`<`) a later calendar date.
* **Hierarchical Evaluation**: When comparing two dates, Python evaluates the components from largest to smallest unit: **Year** → **Month** → **Day**.

In your example, Python checks the year (`2026 == 2026`). Since they match, it checks the month. Because Month `7` (July) is smaller than Month `8` (August), Python immediately determines that `date1` is less than `date2`, returning `True`.


## Understanding UTC Time: Naive vs. Timezone-Aware

When working with dates and times in Python, objects are categorized in two ways:

* **Naive Objects (Old Way)**: Contain clock digits but have **no timezone metadata**. It is like a clock on a wall with no city label. You do not know if `10:00 AM` represents London, New York, or Bengaluru.
* **Timezone-Aware Objects (Modern Way)**: Contain clock digits **plus explicit timezone offsets**. It explicitly states `10:00 AM UTC+00:00`.

### Real-World Use Cases
* **Cloud Logs**: Synchronizes application timelines across servers in different countries (e.g., US vs. India).
* **Databases**: Matches PostgreSQL `TIMESTAMPTZ` columns perfectly, preventing silent data corruption.
* **Calculations**: Prevents runtime crashes (`TypeError`) when comparing local user actions against server timelines.


````python
from datetime import datetime, timezone

# 1. The Deprecated Way (Avoid): Generates a timezone-blind (naive) object
naive_utc = datetime.utcnow()
print("--- Old Method (Naive) ---")
print("Value: ", naive_utc)
print("Timezone Info Field (tzinfo):", naive_utc.tzinfo)

# 2. The Modern Way (Preferred): Generates a timezone-aware object
aware_utc = datetime.now(timezone.utc)
print("\n--- Modern Method (Timezone-Aware) ---")
print("Value: ", aware_utc)
print("Timezone Info Field (tzinfo):", aware_utc.tzinfo)
````

## Updating to the Absolute Latest Standard Syntax

Python's internal warning system recommends using **`datetime.now(timezone.utc)`** to remain fully compatible with upcoming core runtime updates. Let's see how the clean code looks without using any deprecated hooks.


````python
from datetime import datetime, timezone

# The absolute safest, warning-free production standard method
clean_aware_utc = datetime.now(timezone.utc)

print("--- Modern Warning-Free Execution ---")
print("Timestamp:", clean_aware_utc)
print("Timezone: ", clean_aware_utc.tzinfo)
````

## Safe vs. Dangerous Comparison Math

Because naive objects do not contain timezone metadata, Python cannot safely calculate structural time deltas or run logical comparisons between aware and naive objects.


````python
# Demonstrating why mixing object types causes a runtime failure
try:
    # Attempting to compare an aware server timestamp against a naive user timestamp
    print(aware_utc > naive_utc)
except TypeError as error:
    print("❌ Critical System Error Caught:")
    print(f"   -> {error}")
    print("\n💡 Fix: Always use 'datetime.now(timezone.utc)' everywhere to keep code safe.")
````

## Formatting Dates

Raw output from a datetime object looks like `2026-07-26 15:30:12`. Depending on your project requirements, you may need a more human-readable format such as `26/07/2026` or `July 26, 2026`.

### `strftime()`
The **`strftime()`** method stands for **String Format Time**.

**Syntax**:
```python
date.strftime(format)
```

### Formatting Examples
```python
# Format as day/month/year
today.strftime("%d/%m/%Y")  # Output: 26/07/2026

# Format using dashes
today.strftime("%d-%m-%Y")  # Output: 26-07-2026

# Full month name
today.strftime("%B")        # Output: July

# Short month name
today.strftime("%b")        # Output: Jul

# Full weekday name
today.strftime("%A")        # Output: Sunday

# Short weekday name
today.strftime("%a")        # Output: Sun
```


### Most Important Format Codes

These are the codes you will use 95% of the time in data pipelines:

| Code | Meaning |
| :--- | :--- |
| **`%Y`** | 4-digit year |
| **`%y`** | 2-digit year |
| **`%m`** | Month number |
| **`%d`** | Day of the month |
| **`%H`** | Hour (24-hour clock) |
| **`%I`** | Hour (12-hour clock) |
| **`%M`** | Minutes |
| **`%S`** | Seconds |
| **`%A`** | Full weekday name |
| **`%a`** | Short weekday name |
| **`%B`** | Full month name |
| **`%b`** | Short month name |


## Parsing Dates

Suppose an incoming CSV file contains the raw text string `"26-07-2026"`. Because this data is a plain string, Python cannot natively perform date arithmetic or sorting. You must convert it.

### `strptime()`
The **`strptime()`** method stands for **String Parse Time**.

### Parsing Examples
```python
from datetime import datetime

# Example 1: Parsing dashes
date_str = "26-07-2026"
dt = datetime.strptime(date_str, "%d-%m-%Y")
print(dt)  # Output: 2026-07-26 00:00:00

# Example 2: Parsing slashes
datetime.strptime("2026/07/26", "%Y/%m/%d")
```

## Combining Date and Time

You can programmatically merge an isolated date object and an isolated time object into a single unified datetime object.

```python
from datetime import date, time, datetime

d = date.today()
t = time(14, 30)
dt = datetime.combine(d, t)
print(dt)
```
