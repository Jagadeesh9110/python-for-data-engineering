# `collections` Module

The `collections` module is part of Python's Standard Library.

It provides specialized container data structures that extend the capabilities of Python's built-in types like:

- list
- tuple
- dict
- set

These data structures make many common programming tasks simpler, faster, and more readable.

## Why Do We Need collections?

Python already has:
* List
* Tuple
* Dictionary
* Set

So why was another module created?

Because certain problems occur so frequently that Python provides optimized data structures for them.

### Examples

| Problem | Solution |
| :--- | :--- |
| Count occurrences | **Counter** |
| Dictionary with default values | **defaultdict** |
| Fast insert/delete at both ends | **deque** |
| Lightweight object with named fields | **namedtuple** |


## Part 1 — Counter

### What is Counter?
`Counter` counts how many times each element appears in an iterable. Instead of writing counting logic manually, Python does it automatically.


### Without Counter
Suppose you have a list of items and want to calculate their frequency using standard loops:

```python
fruits = ["apple", "banana", "apple", "orange", "banana", "apple"]

count = {}
for fruit in fruits:
    if fruit in count:
        count[fruit] += 1
    else:
        count[fruit] = 1

print(count)
# Output: {'apple': 3, 'banana': 2, 'orange': 1}
```

This approach requires lots of code for a simple operation.


### Using Counter
By importing `Counter` from the `collections` module, you can reduce the entire loop down to a single declaration.

```python
from collections import Counter

fruits = ["apple", "banana", "apple", "orange", "banana", "apple"]
count = Counter(fruits)
print(count)
# Output: Counter({'apple': 3, 'banana': 2, 'orange': 1})
```


````python
from collections import Counter

fruits = ["apple", "banana", "apple", "orange", "banana", "apple"]
count = Counter(fruits)
print(count)
````

### Querying a Counter Instance

* **Getting frequency**: Access counts directly using standard dictionary key lookups.
  ```python
  print(count["apple"])  # Output: 3
  ```

* **Most common values**: Retrieve elements ordered by their frequencies.
  ```python
  count.most_common()
  # Output: [('apple', 3), ('banana', 2), ('orange', 1)]
  ```

* **Filtering for Top N**: Limit the output to return only the highest frequencies.
  ```python
  count.most_common(2)
  # Output: [('apple', 3), ('banana', 2)]
  ```


### Data Engineering Example
Suppose an incoming CSV dataset contains a column tracking corporate departments: `HR`, `IT`, `HR`, `Finance`, `IT`, `HR`. You can calculate department frequencies automatically, a task that is very common in analytics pipelines.

```python
from collections import Counter

departments = ["HR", "IT", "HR", "Finance", "IT", "HR"]
Counter(departments)
# Output: Counter({'HR': 3, 'IT': 2, 'Finance': 1})
```


````python
from collections import Counter

departments = ["HR", "IT", "HR", "Finance", "IT", "HR"]
Counter(departments)
````

````python
from collections import Counter

# Define test payload matching the curriculum context
fruits = ["apple", "banana", "apple", "orange", "banana", "apple"]
count = Counter(fruits)

print("--- Counter String Representation ---")
print(count)

print("\n--- Key Lookup ---")
print("Frequency of apple:", count["apple"])

print("\n--- Frequency Rankings ---")
print("All Common: ", count.most_common())
print("Top 2 Only:", count.most_common(2))
````

## Part 2 — defaultdict

### The Problem with Normal Dictionaries
Attempting to modify a key that does not exist yet inside a normal dictionary will crash your script:

```python
scores = {}
scores["Alice"] += 10
# Output: KeyError: 'Alice'
```

### The Solution: Using `defaultdict`
By importing `defaultdict` from the `collections` module, you can perform modifications on non-existent keys without raising a `KeyError`.

```python
from collections import defaultdict

scores = defaultdict(int)
scores["Alice"] += 10
print(scores)
# Output: defaultdict(<class 'int'>, {'Alice': 10})
```

### Why Does This Work?
Calling `int()` with no arguments returns `0`. When you reference a missing key inside `defaultdict(int)`, Python automatically performs these operations under the hood:
1. Detects that the key `"Alice"` is missing.
2. Initializes `"Alice"` with a default value of `0`.
3. Executes the addition (`0 + 10`), successfully updating the value to `10`.


### Advanced Grouping Operations
A frequent data engineering pattern involves grouping flat record lists into structured category lists. 

Suppose you have raw tabular data:
```python
employees = [
    ("HR", "Alice"),
    ("HR", "Bob"),
    ("IT", "Charlie"),
    ("IT", "David")
]
```

Instead of writing complex nested conditional checks to generate a grouped dictionary, you can pass the Python built-in `list` type as the factory default configuration:

```python
from collections import defaultdict

groups = defaultdict(list)
for dept, name in employees:
    groups[dept].append(name)

print(groups)
# Output: defaultdict(<class 'list'>, {'HR': ['Alice', 'Bob'], 'IT': ['Charlie', 'David']})
```
This is one of the most common structural use cases for `defaultdict` in data analysis pipelines.


## Part 3 — deque

### Understanding List Performance Bottlenecks
Standard Python lists are optimized for appending and popping items at the right-hand end. However, removing an element from the front of a list is a slow operation:

```python
numbers = [1, 2, 3, 4]
numbers.pop(0)
```
When `pop(0)` executes, Python must shift every remaining element left by one slot in memory. For large datasets, this creates a performance bottleneck.

### Optimized Double-Ended Queues
A `deque` (Double-Ended Queue) provides fast, memory-efficient insertion and deletion operations at both ends of the collection.

```python
from collections import deque

queue = deque([1, 2, 3])
queue.append(4)      # Adds to the right side
queue.appendleft(0)  # Adds to the left side
print(queue)
# Output: deque([0, 1, 2, 3, 4])
```

### Item Removal Mechanics
```python
queue.pop()      # Removes and returns from the right side
queue.popleft()  # Removes and returns from the left side
```
Both push and pop modifications run with optimized efficiency, making `deque` ideal for streaming queues and sliding window calculation algorithms.


## Part 4 — namedtuple

### The Readability Issue with Standard Tuples
Using standard numeric tuples to represent structured data rows forces you to reference parameters using cryptic positional indices, which makes your code difficult to read:

```python
employee = (101, "Alice", 65000)
print(employee[1])  # Returns "Alice" but context is unclear
```

### Creating Lightweight Objects with Named Fields
A `namedtuple` assigns meaningful property names to each tuple index position. This grants your data structures clean, object-like dot-notation access while preserving the low memory overhead of regular tuples.

```python
from collections import namedtuple

# Define the structured schema type
Employee = namedtuple("Employee", ["id", "name", "salary"])

# Instantiate data instances
emp = Employee(101, "Alice", 65000)

print(emp.name)    # Output: Alice
print(emp.salary)  # Output: 65000
```
Using field attributes makes your script significantly clearer than relying on numeric index positions.

| **`namedtuple`** | Assigns explicit named fields to tuple records. |


````python
from collections import defaultdict, deque, namedtuple

# 1. Verify defaultdict grouping operation
print("--- 1. defaultdict Grouping Execution ---")
employees = [("HR", "Alice"), ("HR", "Bob"), ("IT", "Charlie"), ("IT", "David")]
groups = defaultdict(list)
for dept, name in employees:
    groups[dept].append(name)
print("Grouped Dictionary output:", dict(groups))

# 2. Verify deque manipulation speeds
print("\n--- 2. deque Fast Boundary Pushes ---")
stream_queue = deque([10, 20, 30])
stream_queue.appendleft(5)
stream_queue.append(40)
print("Updated Queue structure: ", stream_queue)
stream_queue.popleft()
print("Post-popleft structure:  ", stream_queue)

# 3. Verify namedtuple reference mapping
print("\n--- 3. namedtuple Attribute References ---")
LogRecord = namedtuple("LogRecord", ["timestamp", "severity", "message"])
log_row = LogRecord("2026-07-26", "ERROR", "PostgreSQL database handshake failed.")
print(f"Log Details -> Level: {log_row.severity} | Info: {log_row.message}")
````

## defaultdict - Group employees by department

````python
from collections import defaultdict

# Flat dataset layout typical of database query rows
raw_employee_rows = [
    ("Engineering", "Alice"),
    ("Sales", "Bob"),
    ("Engineering", "Charlie"),
    ("Marketing", "David"),
    ("Sales", "Eve")
]

# Initialize a dictionary that creates empty lists for new keys
department_groups = defaultdict(list)

for department, employee_name in raw_employee_rows:
    department_groups[department].append(employee_name)

print("--- defaultdict: Department Grouping ---")
for dept, roster in department_groups.items():
    print(f"Department: {dept:12} | Roster: {roster}")
````

## deque - Maintain the latest 100 log records in memory

````python
from collections import deque

# Maxlen automatically discards old entries when new ones are appended left/right
log_stream_buffer = deque(maxlen=5) # Kept at 5 for clear visualization output

print("--- deque: Rolling Memory Log Buffer ---")
# Simulating a stream of incoming server status logs
for log_id in range(1, 9):
    log_line = f"2026-07-26 18:18:00 - INFO - Processing item #{log_id}"
    log_stream_buffer.append(log_line)
    print(f"Pushed log #{log_id}. Current buffer size: {len(log_stream_buffer)}")

print("\nFinal items remaining in active memory:")
for log in log_stream_buffer:
    print(f" • {log}")
````

## namedtuple - Represent rows from a CSV with named attributes

````python
from collections import namedtuple

# 1. Define the structural schema representing CSV headers
CSVRowSchema = namedtuple("CSVRowSchema", ["id", "product_name", "unit_price", "inventory_count"])

# 2. Simulate streaming parsed rows from an external data pipeline
mock_csv_file_stream = [
    ("1024", "Wireless Mouse", "24.99", "150"),
    ("1025", "Mechanical Keyboard", "89.99", "45"),
    ("1026", "Type-C Hub Adapter", "19.50", "300")
]

print("--- namedtuple: Object-Mapped Tabular Records ---")
for raw_tuple in mock_csv_file_stream:
    # Unpack the raw data row directly into our schema template
    row = CSVRowSchema(*raw_tuple)
    
    # Access fields explicitly by name instead of using unreadable numbers like row[1]
    total_value = float(row.unit_price) * int(row.inventory_count)
    print(f"Item ID: {row.id} | Product: {row.product_name:20} | Stock Value: ${total_value:.2f}")
````
